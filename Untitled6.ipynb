```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import tensorflow as tf
import string
import gc

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, Dense

print("TensorFlow:", tf.__version__)
```

```python
df = pd.read_csv("Qoute_dataset.csv")

print(df.head())
print(df.shape)
print(df.columns)
```

```python
quotes = df['quote'].dropna().astype(str)

quotes = quotes.str.lower()

translator = str.maketrans('', '', string.punctuation)

quotes = quotes.apply(lambda x: x.translate(translator))

print(quotes.head())
```

```python
vocab_size = 10000

tokenizer = Tokenizer(num_words=vocab_size)

tokenizer.fit_on_texts(quotes)

word_index = tokenizer.word_index

print("Vocabulary:", len(word_index))

print(list(word_index.items())[:10])
```

```python
sequence = tokenizer.texts_to_sequences(quotes)

for i in range(3):
    print(quotes.iloc[i])
    print(sequence[i])
```

```python
X = []
y = []

for seq in sequence:
    for i in range(len(seq)):
        input_seq = seq[:i]
        output_seq = seq[i]

        X.append(input_seq)
        y.append(output_seq)

print("Total samples:", len(X))
print("Total labels:", len(y))
```

```python
max_len = max(len(x) for x in X)

print("Maximum length:", max_len)

X_padded = pad_sequences(
    X,
    maxlen=max_len,
    padding='pre',
    dtype='int32'
)

print("X shape:", X_padded.shape)
```

```python
y = np.array(y, dtype=np.int32)

print("Y shape:", y.shape)
print("Y dtype:", y.dtype)
```

```python
del X
del sequence

gc.collect()
```

```python
print("X RAM:",
      X_padded.nbytes / (1024**2), "MB")

print("Y RAM:",
      y.nbytes / (1024**2), "MB")
```

```python
embedding_dim = 50
rnn_units = 128

rnn_model = Sequential()

rnn_model.add(
    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim,
        input_length=max_len
    )
)

rnn_model.add(
    SimpleRNN(units=rnn_units)
)

rnn_model.add(
    Dense(
        units=vocab_size,
        activation='softmax'
    )
)

rnn_model.summary()
```

```python
rnn_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
epochs = 10
batch_size = 128

history_rnn = rnn_model.fit(
    X_padded,
    y,
    epochs=epochs,
    batch_size=batch_size
)
```

```python
lstm_model = Sequential()

lstm_model.add(
    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim,
        input_length=max_len
    )
)

lstm_model.add(
    LSTM(units=rnn_units)
)

lstm_model.add(
    Dense(units=vocab_size, activation='softmax')
)

lstm_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

lstm_model.summary()
```

```python
history_lstm = lstm_model.fit(
    X_padded,
    y,
    epochs=100,
    batch_size=128,
    validation_split=0.1
)
```

```python
lstm_model.save("lstm_model.h5")
```

```python
index_to_word = {}
for word, index in word_index.items():
    index_to_word[index] = word
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
def predictor(model,tokenizer,text,max_len):
  text = text.lower()
  seq = tokenizer.texts_to_sequences([text])[0]
  seq = pad_sequences([seq],maxlen=max_len,padding='pre')
  pred = model.predict(seq,verbose=0)
  pred_indx=np.argmax(pred)
  return index_to_word.get(pred_indx,"")
```

```python
seed_text="what are you"
next_word = predictor(lstm_model,tokenizer,seed_text,max_len)
print(next_word)
```

```python
def generate_text(model, tokenizer, max_len, seed_text, n_words):
    for _ in range(n_words):
        next_word = predictor(model, tokenizer, seed_text, max_len)
        if(next_word==""):
            break
        seed_text += " " + next_word
    return seed_text
```

```python
seed = "the meaning of life"
generate_text=generate_text(lstm_model,tokenizer,max_len,seed,10)
print(generate_text)
```
